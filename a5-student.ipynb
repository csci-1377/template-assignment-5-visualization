# Part A. Datasets

## Dunkin' Donuts Across the US

For this assignment, we'll be using two datasets. The first is a dataset of about Dunkin' Donuts storefronts across the United States.

```python
dunkins = a5.load_dunkins()
dunkins
```

This dataset contains seven columns:
* `state`: two-letter abbreviation of a US state.
* `region`: the region of the country the US state is in, e.g., Northeast for RI.
* `store_count`: the number of stores in the given state.
* `stores_per_100k`: the number of stores per 100,000 residents of the state.
* `stores_per_sqmi`: the number of stores per square mile in the state.
* `avg_sat_delivery_hours`: the average number of hours delivery is available on Saturdays.
* `frac_loyalty`: the fraction of storefronts that support the Dunkins loyalty program.

## Life Expectancy in Rhode Island

The second dataset provides county-level statistics about life expectancy and potential correlates in Rhode Island.

```python
life = a5.load_ri_life()
life
```

This dataset contains ten columns:
- `county`: The name of the county in Rhode Island.
- `geometry`: a description of the geometry of the county, used for map visualizations.
- `rural_urban`: a statistical classification of whether a county is more rural or urban.
- `life_expectancy`: the average life expectancy in the county.
- `asmr`: age-standardized mortality rate (ASMR) per 100,000, or the number of people who die every year, weighted to enable comparisons across different age distributions.
- `asmr_rank`: the ASMR rank across all counties (1 is the lowest ASMR).
- `rent`: The median rent in the county.
- `gini`: The [Gini coefficient](https://en.wikipedia.org/wiki/Gini_coefficient), a measure of wealth inequality in the county.
- `pct_food_desert`: The percentage of the population in the county that lives in a USDA-defined [food desert](https://en.wikipedia.org/wiki/Food_desert).
- `fast_food`: The number of fast food and convenience stores per square mile in the county.
- `pct_farmers_market`: The percentage of the population that lives within 10 minutes of a farmer's market.

## Task 1: Is Rhode Island the Dunkin's capital of the world? (10 points)

This task focuses on the `state`, `store_count`, `stores_per_sqmi`, and `stores_per_100k` variables to analyze whether Rhode Island can claim to be the Dunkin's capital of the world.

### Chart 1A

```python
base = alt.Chart(dunkins).encode(x="state:N")
base.mark_point(filled=True).encode(size="store_count", color="stores_per_sqmi") + \
base.mark_point(filled=False).encode(size="store_count", strokeWidth="stores_per_100k")
```

### Chart 1B

```python
base.mark_bar().encode(y="stores_per_100k:Q", size="store_count:Q", color="region") + \
base.mark_point().encode(y="stores_per_100k:Q", yOffset=alt.value(-10), opacity="stores_per_sqmi")
```

### Chart 1C

```python
graph3 = alt.Chart(dunkins).encode(y="state:N")
(
    graph3.mark_point(filled=False).encode(x="stores_per_100k", size=alt.value(260), strokeWidth=alt.value(3), color=alt.value("#111")) + \
    graph3.mark_point(filled=True).encode(x="stores_per_100k", size=alt.value(90), color=alt.value("white")) + \
    graph3.mark_point(filled=True).encode(x="stores_per_100k", xOffset=alt.value(20), opacity=alt.value(0.7), size="stores_per_sqmi")
).properties(width=800, height=450)
```

## Task 2: How does housing affordability relate to ASMR? (10 points)

This tasks focuses on the `asmr`, `rent`, `gini`, and `rural_urban` variables to analyze how the age-standardized mortality rate relates to a county's rent and Gini coefficient, considering the distinction between rural and urban counties.

### Chart 2A

```python
chart = alt.Chart(life)
chart.mark_point().encode(
    x=alt.X("rent", scale=alt.Scale(zero=False)),
    y=alt.Y("asmr", scale=alt.Scale(zero=False)),
    size="gini",
    shape="rural_urban"
)
```

### Chart 2B

```python
chart.mark_bar().encode(x="rent", y="asmr", row="rural_urban", color="gini")
```

### Chart 2C

```python
(
    chart.mark_line().encode(x="asmr_rank", y=alt.Y("rent", scale=alt.Scale(zero=False)), color="rural_urban") &
    chart.mark_line().encode(x="asmr_rank", y=alt.Y("gini", scale=alt.Scale(zero=False)), color="rural_urban")
)
```

## Task 3: How does life expectancy vary across Rhode Island? (10 points)

This task focuses on the `county`, `life_expectancy`, and `rural_urban` variables to analyze how life expectancy changes county-by-county, with an aggregate focus on comparing rural vs. urban counties.

### Chart 3A

```python
alt.Chart(life).mark_bar().encode(
    x=alt.X("county", sort="-y"), 
    y=alt.Y("life_expectancy:Q", scale=alt.Scale(domainMin=75, clamp=True)), 
    color="rural_urban"
)
```

### Chart 3B

Note that this chart contains a tooltip &mdash; try hovering your mouse over each county.

```python
hover = alt.selection_point(on='mouseover', empty=False)
geo_base = alt.Chart(life).mark_geoshape() \
    .project(type="identity", reflectY=True) \
    .encode(strokeWidth=alt.condition(hover, alt.value(5.), alt.value(1.)), stroke=alt.value("#ccc")).add_params(hover) \
    .properties(width=600, height=600)

geo_base.encode(color="life_expectancy", tooltip=["county", "life_expectancy", "rural_urban"])
```

### Chart 3C

Note that this chart contains a selection interval &mdash; click and drag your mouse within the bar chart below.

```python
brush = alt.selection_interval(encodings=['x'])
geo_base.encode(
    color="rural_urban", 
    tooltip=["county", "life_expectancy"], 
    opacity=alt.condition(brush, alt.value(1.0), alt.value(0.3))
) & \
alt.Chart(life).mark_bar().encode(
    x=alt.X("county:N", sort='-y'), 
    y=alt.Y("life_expectancy:Q", scale=alt.Scale(domainMin=75, clamp=True))
).add_params(brush)
```
